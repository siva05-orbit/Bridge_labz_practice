```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread(r"C:\Users\Lenovo\Downloads\images.jpg")
print(type(img))
```

```python
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(rgb); plt.axis('off')
```

## Image Properties

```python
print('Shape:',img.shape)
print('Size:',img.size)
print('Datatype:',img.dtype)
```

```python
resized=cv2.resize(img,(300,300))
plt.imshow(cv2.cvtColor(resized,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

## Crop

```python
crop=img[50:250,50:250]
plt.imshow(cv2.cvtColor(crop,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

```python
h,w=img.shape[:2]
M=cv2.getRotationMatrix2D((w//2,h//2),45,1)
rot=cv2.warpAffine(img,M,(w,h))
plt.imshow(cv2.cvtColor(rot,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

```python
flip=cv2.flip(img,1)
plt.imshow(cv2.cvtColor(flip,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
plt.imshow(gray,cmap='gray'); plt.axis('off')
```
