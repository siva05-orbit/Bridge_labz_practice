```python
import cv2
import matplotlib.pyplot as plt
img=cv2.imread('image.jpg')
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

## Thresholding

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
_,th=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
plt.imshow(th,cmap='gray'); plt.axis('off')
```

## Gaussian Blur

```python
g=cv2.GaussianBlur(img,(5,5),0)
plt.imshow(cv2.cvtColor(g,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

```python
m=cv2.medianBlur(img,5)
plt.imshow(cv2.cvtColor(m,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

```python
b=cv2.bilateralFilter(img,9,75,75)
plt.imshow(cv2.cvtColor(b,cv2.COLOR_BGR2RGB)); plt.axis('off')
```

```python
sx=cv2.Sobel(gray,cv2.CV_64F,1,0)
plt.imshow(sx,cmap='gray'); plt.axis('off')
```

## Canny Edge Detection

```python
edges=cv2.Canny(gray,100,200)
plt.imshow(edges,cmap='gray'); plt.axis('off')
```

## Morphological Operations

```python
kernel=cv2.getStructuringElement(cv2.MORPH_RECT,(5,5))
erode=cv2.erode(th,kernel)
dilate=cv2.dilate(th,kernel)
open_img=cv2.morphologyEx(th,cv2.MORPH_OPEN,kernel)
close_img=cv2.morphologyEx(th,cv2.MORPH_CLOSE,kernel)
print('Done')
```
