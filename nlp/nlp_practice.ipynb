```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

```python
text = "Conor McGregor suffered a severe right knee injury—suspected to be a blown ACL or torn meniscus—just 69 seconds into his return fight against Max Holloway at UFC 329. The injury occurred instantly when McGregorv landed awkwardly after attempting an opening jumping roundhouse kick, causing him to repeatedly lose his footing before the referee stopped the bout. "
```

```python
sentences = sent_tokenize(text)

for s in sentences:
    print(s)
```

```python
words = word_tokenize(text)

for i in words:
    print(i)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = []

for word in words:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

```python
from nltk.corpus import gutenberg

nltk.download('gutenberg')

print(gutenberg.fileids())
```

```python
corpus = gutenberg.raw('austen-emma.txt')

print(corpus[:1000])
```

```python
from nltk import FreqDist

tokens = word_tokenize(corpus.lower())

fdist = FreqDist(tokens)

print(fdist.most_common(10))
```

```python
print("Total Words :",len(tokens))
print("Unique Words :",len(set(tokens)))
```

```python
#word embeddings

sentences = [
    ["apple","iphone","mobile"],
    ["samsung","galaxy","phone"],
    ["apple","macbook","laptop"],
    ["hp","laptop","computer"],
    ["dell","computer","laptop"],
    ["iphone","camera","apple"],
    ["galaxy","camera","samsung"]
]

from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=100,
    window=3,
    min_count=1,
    workers=4
)
```

```python
model.wv["apple"]
```

```python
model.wv.most_similar("apple")
```

```python
model.wv.similarity("apple","iphone")
```
