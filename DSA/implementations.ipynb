```python
#STACK IMPLEMENTATION
class Stack:
    def __init__(self):
        self.items=[]

    def empty(self):
        return self.items

    def add(self,data):
        self.items.append(data)

    def remove(self):
        self.items.pop()
    def stack(self):
        return self.items

if __name__ == "__main__":
    q = Stack()

    q.empty()
    q.add(20)
    q.add(10)
    q.add(40)
    q.remove()
    print(q.stack())
```

```python
#QUEUE IMPLEMENTATION

class Queue:

    def __init__(self):
        self.items=[]
    
    def add(self,data):
        self.items.append(data)
    def remove(self):
        self.items.pop(0)
    def queue(self):
        print(self.items)
        print(len(self.items))
    

if __name__== '__main__':
    q=Queue()

    q.add(3)
    q.add(2)
    q.add(9)
    q.add(10)
    q.remove()
    q.queue()
```

```python
#LINKED LIST IMPLEMENTATION
class Node():
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
        
class LinkedList():
    def __init__(self):
        self.head = None
    
    def add(self,data):
        node=Node(data)
        if self.head == None:
            self.head=node
        else:
            node.next=self.head
            node.next.prev = node
            self.head = node
        
    def search(self,element):
        obj = self.head
        if obj.data == element:
            return obj
        while obj.next is not None:
            if obj.data == element:
                return obj
            obj=obj.next
        
        return None
    
    def remove(self,p):
        if p.prev is not None:
            p.prev.next = p.next
        if p.next is not None:
            p.next.prev=p.prev

if __name__=="__main__":
    ll = LinkedList()

    ll.add(10)
    ll.add(20)
    ll.add(30)

    print("Searching for 20...")
    node = ll.search(20)

    if node:
        print("Found:", node.data)
    else:
        print("Not found")

    print("Removing 20...")
    ll.remove(node)

    node = ll.search(20)
    if node:
        print("Still exists")
    else:
        print("Removed successfully")

        
```
